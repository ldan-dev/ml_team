# Clasificación del dataset Iris
Pasos según el pizarrón:
1. Cargar el conjunto de datos Iris plant
2. Dividir de forma estratificada: 80% entrenamiento / 20% prueba
3. Usar Nearest Centroid (mínima distancia) con Minkowski p=2
4. Usar K-Nearest Neighbors (K vecinos más cercanos), con k ∈ {1, 3, 5}
5. Calcular accuracy en entrenamiento y en prueba

## 1. Cargar el conjunto de datos Iris plant

```python
from sklearn.datasets import load_iris

# cargar el conjunto de datos iris
X, Y = load_iris(return_X_y=True)

print("forma de X:", X.shape)
print("forma de Y:", Y.shape)
```

## 2. Dividir de forma estratificada (80% entrenamiento / 20% prueba)
Usamos `stratify=Y` para que la proporción de cada clase (setosa, versicolor, virginica) se mantenga igual en ambos conjuntos.

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    X,
    Y,
    test_size=0.2,      # 80% entrenamiento, 20% prueba
    stratify=Y,          # división estratificada por clase
    random_state=432      # reproducibilidad
)

print("muestras de entrenamiento:", len(x_train))
print("muestras de prueba:", len(x_test))
```

## 3. Nearest Centroid (mínima distancia) con Minkowski p=2
`NearestCentroid` calcula el centroide de cada clase y clasifica cada muestra según el centroide más cercano.
Minkowski con `p=2` equivale a la distancia euclidiana.

```python
from sklearn.neighbors import NearestCentroid
from sklearn.metrics import accuracy_score

# Nearest Centroid con distancia Minkowski p=2 (equivalente a euclidiana)
nc_model = NearestCentroid(metric="euclidean")  # minkowski con p=2
nc_model.fit(x_train, y_train)

# predicciones
y_train_pred_nc = nc_model.predict(x_train)
y_test_pred_nc = nc_model.predict(x_test)

# accuracy
acc_train_nc = accuracy_score(y_train, y_train_pred_nc)
acc_test_nc = accuracy_score(y_test, y_test_pred_nc)

print("Nearest Centroid")
print("accuracy entrenamiento:", acc_train_nc)
print("accuracy prueba:", acc_test_nc)
```

## 4. K-Nearest Neighbors con k ∈ {1, 3, 5}
`KNeighborsClassifier` usa por defecto la distancia Minkowski con `p=2` (euclidiana), así que no hace falta cambiar nada extra.

```python
from sklearn.neighbors import KNeighborsClassifier

resultados_knn = []

for k in [1, 3, 5]:
    knn_model = KNeighborsClassifier(
        n_neighbors=k,
        metric="minkowski",
        p=2  # p=2 -> distancia euclidiana
    )
    knn_model.fit(x_train, y_train)

    y_train_pred = knn_model.predict(x_train)
    y_test_pred = knn_model.predict(x_test)

    acc_train = accuracy_score(y_train, y_train_pred)
    acc_test = accuracy_score(y_test, y_test_pred)

    resultados_knn.append((k, acc_train, acc_test))

    print(f"K = {k}")
    print("accuracy entrenamiento:", acc_train)
    print("accuracy prueba:", acc_test)
    print("-------------")
```

## 5. Resumen de resultados

```python
import pandas as pd

resumen = pd.DataFrame(
    [("NearestCentroid", "-", acc_train_nc, acc_test_nc)] +
    [("KNN", k, acc_train, acc_test) for k, acc_train, acc_test in resultados_knn],
    columns=["modelo", "k", "accuracy_entrenamiento", "accuracy_prueba"]
)
resumen
```
